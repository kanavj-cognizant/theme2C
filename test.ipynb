```python
import whitebox_workflows
from whitebox_workflows import WbEnvironment
```

```python
from whitebox_workflows import WbEnvironment, download_sample_data

wbe = WbEnvironment()
print(wbe.version())  # Print the version number
```

```python
wbe.verbose = True
wbe.working_directory = download_sample_data('mill_brook')
print(f'Data have been stored in: {wbe.working_directory}')
```

```python
lidar = wbe.read_lidar('mill_brook.laz')
print(
    f"There are {lidar.header.number_of_points} points in the lidar dataset.")
```

```python
# Create a DEM
dem = wbe.lidar_tin_gridding(lidar, returns_included='all', cell_size=1.0, excluded_classes=[
                             1], max_triangle_edge_length=100.0)

# Fill missing data
# Notice that we overwrite `dem` here.
dem = wbe.fill_missing_data(dem, filter_size=35)
# You can save the smoothed DEM if you like.
# wbe.write_raster(dem, 'dem.tif', compress=False)

# Let's visualize the DEM with a hillshade
hs = wbe.multidirectional_hillshade(dem)
# Compression is good, but it is a bit slower so here we won't use it.
wbe.write_raster(hs, 'hillshade.tif', compress=False)
```

```python
dem_smoothed = wbe.feature_preserving_smoothing(
    dem, filter_size=11, normal_diff_threshold=25.0, iterations=3)
hs = wbe.multidirectional_hillshade(dem_smoothed)
wbe.write_raster(hs, 'hillshade_smoothed.tif', compress=False)
```

```python
contours = wbe.contours_from_raster(dem_smoothed, contour_interval=10.0)
wbe.write_vector(contours, 'contours.shp')
```

```python
import elevatr
import whitebox_workflows
from whitebox_workflows import WbEnvironment
```

```python
extent = 0.05
rs = elevatr.get_elev_raster((-1.75926729-extent, 54.97627333-extent, -1.52526845+extent, 55.04531808+extent), zoom=12,use_cache=True, delete_cache=False,crs="EPSG:4326")
```

```python
rs.to_tif("./wbe/raster.tif")
```

```python
wbe = whitebox_workflows.WbEnvironment()
wbe.verbose = False
```

```python
wbe.working_directory = "./wbe"
```

```python
rs = wbe.read_raster("raster.tif")
```

```python
# Change the max dist parameter as appropriate for your DEM
# filled = wbe.breach_depressions_least_cost(
#     rs, flat_increment=0.001, max_dist=100)
filled = wbe.fill_depressions(rs)
# wbe.write_raster(filled,"raster_filled.tif")
```

```python
d8s = wbe.d8_pointer(filled)
# wbe.write_raster(d8s, "raster_d8s.tif")
```

```python
accum = wbe.d8_flow_accum(d8s,out_type="cells",clip=False,input_is_pointer=True)
# wbe.write_raster(accum, "raster_flow_accum.tif")
```

```python
streams = wbe.extract_streams(accum,threshold=1000)
```

```python
points = wbe.read_vector("riskmap.shp")
```

```python
snapped = wbe.jenson_snap_pour_points(points,streams,snap_dist=0.001)
```

```python
watersheds = wbe.watershed(d8_pointer=d8s,pour_points=snapped,esri_pntr=False)
```

```python
wbe.write_raster(watersheds,"dilineated.tif")
```

```python
polygons = wbe.raster_to_vector_polygons(watersheds)
```

```python
wbe.write_vector(polygons,"pourpoints_watershed.shp")
```
